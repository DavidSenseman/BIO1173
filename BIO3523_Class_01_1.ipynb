## **How Hermes Agent Works**

Hermes Agent is one type of an **LLM harness**.  An LLM harness acts as the _operating system_ surrounding a large language model (LLM). Without a harness, a LLM is basically "a brain in a jar". The LLM (e.g. "ChatGPT") can think, but it can't really **_do_** anything with its thoughts. An AI agent, like Hermes Agent, **provides a body** for the brain, converting thoughts into physical actions.

The software underlying Hermes Agent has the following 4 parts:
1. **Execution Loop**
2. **Context Routing**
3. **Tool Parsing**
4. **Memory Persistence**

Together, these 4 parts allow any foundation model to function as your long-running, personal assistant.

#### **The Core Agent Loop (Observation-Action Cycle)**

At its lowest level, the harness manages the event loop between the user, the environment, and the model:

1. **Input Interception:** Incoming messages from any interface (CLI, Telegram, Discord, Slack) are normalized into standard user turns.
2. **Context Assembly:** Before hitting the LLM, the harness injects relevant system prompts, active project context files, persistent user memory, and available tool definitions.
3. **Inference & Tool Parsing:** The LLM receives the prompt and emits either direct prose or structured tool calls. Hermes Agent intercepts these tool requests, executes them safely, and appends the result back into the prompt context.
4. **Execution Decision:** If the model returns tool calls, the harness recursively loops (running tools and feeding back stdout/stderr) until the model satisfies the objective or requests user input.

---

#### **Dynamic Memory & The "Closed Learning Loop"**

Standard LLM calls forget everything once the context window closes. Hermes Agent acts as a persistent layer by surrounding the LLM with three memory subsystems:

* **Episodic Recall (FTS5 & Vector Search):** Past conversation logs are indexed locally in SQLite. The harness automatically pulls relevant historical snippets into current turns when related topics re-emerge.
* **Agent Nudges:** The harness periodically interrupts the execution stream with system prompts ("nudges") directing the LLM to summarize key user facts, project states, or lessons learned into long-term storage.
* **Self-Generated Skills (`agentskills.io`):** When the LLM successfully completes a complex, multi-step trajectory, the harness prompts it to synthesize the solution into a structured `SKILL.md` file. In future runs, the harness injects these custom skills as available tools.

---

#### **Sandboxed Tool Execution Environments**

An LLM cannot run shell commands or write files directly; it can only output text. The harness acts as the execution proxy:

* **Backend Abstraction:** Hermes routes command execution through configurable backends—ranging from a local subshell to hardened Docker containers, remote SSH nodes, or cloud serverless backends (e.g., Daytona, Modal).
* **Native Tool Integration & MCP:** Out of the box, the harness exposes 40+ native capabilities (web search, browser automation, file I/O) alongside support for Model Context Protocol (MCP) servers.
* **Programmatic Execution (`execute_code`):** To avoid burning token budget on repetitive step-by-step tool turns, the harness allows the LLM to write Python scripts that invoke tools locally via RPC in a single inference cycle.

---

#### **State & Context Management**

Raw LLMs struggle when message histories exceed context limits or token costs spike. The harness handles the scaffolding:

* **Context Compression:** The harness monitors token usage and transparently condenses old execution traces or tool outputs while preserving core facts.
* **Sub-Agent Delegation:** When tasked with large parallel jobs, the main harness can spawn lightweight, isolated child instances with their own execution loops, minimizing context pollution in the main thread.

## **Burning Tokens**

Because Hermes is an autonomous agent framework, every API call carries substantial system overhead (tool definitions, system instructions, memory files, and skill indexes) before the model processes your actual two-word query, "Hi Hermes".

#### **Input Tokens (~12,000 to 16,000 tokens**):
* **Tool Schemas (JSON/XML):** **~8,000 – 10,000 tokens**

## **Using Free Models with Hermes Agent**

While free API tiers (like OpenRouter’s :free models or Hugging Face free inference) seem ideal for a college student budget, agentic frameworks like Hermes Agent operate very differently from standard chat interfaces. Because Hermes relies on continuous tool-calling loops, background memory, and multi-turn subagent tasks, relying on free-tier models introduces severe operational bottlenecks that make development frustrating.

#### **Rapid Rate-Limit Exhaustion (429 Errors)**

* **The Problem:** In a typical chat interface, one user prompt equals one API call. In Hermes Agent, a single user prompt triggers a multi-step execution loop: planning $\rightarrow$ tool call $\rightarrow$ tool output $\rightarrow$ reflection $\rightarrow$ final answer. A single task can generate 5 to 15 API requests under the hood.
* **The Impact:** Free tiers on OpenRouter (typically capped at 20 requests/minute or 200 requests/day) hit rate limits halfway through a single multi-step task. The agent freezes mid-session or crashes with HTTP 429 errors.


#### **Broken Tool & Function Calling ("Talks, But Does Nothing")**

* **The Problem:** Hermes Agent relies on native JSON/XML function calling to execute shell commands, edit files, or browse the web. Many free-tier endpoints route to smaller or unoptimized models that lack robust tool-calling support.
* **The Impact:** Free models often output raw tool JSON or XML tags as standard message text (e.g., <tool_call>{"name": "filesystem"}</tool_call>) instead of structuring it as a system API call. Hermes treats this as plain conversational text, leading to the common issue where the agent talks about creating a file or running a script, but never actually executes it.

#### **Severe Context Window Capping**

* **The Problem:** To keep hosting costs zero, free provider endpoints frequently cap input context windows (often to 4,096 or 8,192 tokens), even if the underlying model architecture supports much more.
*  **The Impact:** Hermes injects substantial system instructions, memory files, skill sets, and tool definitions into every prompt. In a truncated context window, the agent quickly loses track of earlier context, forgets user instructions mid-task, or enters endless repetition loops.

#### **High Latency, Cold Starts, and Unreliability**

* **The Problem:** Free endpoints run on low-priority or queue-backed serverless infrastructure.
* **The Impact:** Response latency can range from 10 to 30 seconds per turn due to cold starts. Furthermore, free endpoints suffer high rates of 502 Bad Gateway or 503 Service Unavailable errors during peak hours, breaking scheduled cron tasks or multi-agent workflows.


Practical Comparison
Feature / Factor | Free API Models (:free suffix) | Low-Cost OpenRouter API Models (e.g., DeepSeek, Qwen 2.5 72B, Llama 3.3)
--- | --- | ---
Tool Calling Reliability | Flaky; often outputs tool calls as plain text | Reliable native JSON function-calling support
Rate Limits | Harsh per-minute / per-day caps | Capped only by account credit balance
Context Window | Capped/truncated by host (4k–8k) | Full context window available (32k–128k+)

## **Can Someone Steal My API Key from the MD5 Hash?**

The short answer is no—an MD5 hash cannot be directly reversed to recover your API key. The **MD5 hash algorithm** is a one-way cryptographic function that transforms any input into a fixed 128-bit digest. The reason you cannot regenerate the original text from this hash is that the process is inherently **lossy**. During hashing, the algorithm compresses the data through a series of bitwise operations and modular additions that discard information, mapping an effectively infinite range of possible inputs onto a finite output space.

Because multiple different inputs can produce the same hash (a collision), and because the original byte sequence is mathematically scrambled and truncated, the specific information required to reconstruct the original text is simply no longer present in the final output. In essence, the hash is a fingerprint of the data, not an encrypted version of it—the transformation destroys the source data, making it computationally infeasible to reverse the process.

**NOTE: Since a particular API key always produces the _same_ hash value, it is easy to tell if two students submit the _same_ API key in this lesson.**

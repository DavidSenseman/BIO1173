####**1. Test the Actual Service Port (Port 53)**
Sometimes network administrators block ICMP (Ping) traffic to servers, but leave the actual service ports open. Since 10.251.53.195 is a DNS server, check if TCP Port 53 is reachable, even if ping fails.

In your PowerShell give this command:
```text
Test-NetConnection -ComputerName 10.251.53.195 -Port 53
```
If `TcpTestSucceeded` returns `True`, the server is online and reachable, but ping is just disabled by a firewall. If it returns `False`, there is a genuine connectivity or routing block.

####**2. Force a DNS Query**
To verify if the DNS server is responding to actual DNS requests (UDP/TCP) rather than just connectivity checks.

In your PowerShell give this command:
```text
Resolve-DnsName -Name utsa.edu -Server 10.251.53.195
```
If this returns an IP address for `utsa.edu`, the DNS server is fully functional and the student has access to it. If it times out, the traffic is being dropped.

####**3. Trace the Network Path**

Identify exactly where the traffic is getting stopped or misrouted.

In your PowerShell give this command:
```text
tracert 10.251.53.195
```
Check the first few hops. If the traffic immediately routes to an external ISP gateway (like a home router or cellular gateway) instead of a UTSA internal router, the computer's routing table is misconfigured. If it reaches a UTSA gateway and then dies, a firewall rule or network segment is blocking the path.

####**4. Check the Routing Table**

Verify how Windows is instructed to handle traffic destined for the 10.x.x.x private network.

In your PowerShell give this command:
```text
Get-NetRoute -DestinationPrefix 10.*
```
Ensure there is a route directing 10.x.x.x traffic to the correct interface (e.g., the UTSA VPN adapter or the physical campus Ethernet adapter). If no specific 10.x.x.x route exists, the computer will send the traffic to the 0.0.0.0 default gateway, which may just drop it if it's external.

####**5. Verify Network Adapter and IP Lease**

Confirm you are actually on the correct network segment (e.g., UTSA Secure Wi-Fi instead of UTSA Guest).

In your PowerShell give this command:
```
ipconfig /all
```
Check the IPv4 Address and Default Gateway. If you are on a guest network, you may have internet access (hence 8.8.8.8 works) but be placed on an isolated VLAN that strictly denies access to internal infrastructure like 10.251.53.195.

# **Cloud-Based Models**

Up until now, your Hermes Agent has been using a local Large Language Model (LLM) as its "brains". A **local model** is a LLM that is running on (or _hosted_ ) by a local computer with one (or more) graphical processing units (GPUs). For technical reasons, each GPU must have a large amount of video ram (VRAM). For example, the default local model for your Hermes Agent is called `Nemotron`. and it is being hosted on a DELL ProMax GB10 with **`128GB of VRAM`**.

While this local model is "smart enough" to able your Hermes Agent to perform all of the basic tasks we have asked it sofar, we will need to switch to a more powerful ("smarter") cloud-based model for the lessons in this module.

###**API Keys**

While local models are free to use, you are charged whenever you use a cloud-based model. The most common way of billing you is through an **API Key (Application Programming Interface Key)**. Your API Key is a unique, secret string of characters that acts as a digital passport for OpenRouter as well as other software applications. When you want a local script or application to interact with a cloud-based service, the API key authenticates the request and authorizes access to the remote server.

In the context of accessing cloud-based AI models, an API key serves three primary functions:

* **Authentication:** It proves your identity to the cloud provider, verifying that the incoming request is coming from a registered and approved user.

* **Billing and Token Tracking:** Cloud-based inference requires significant computational resources, so providers charge based on usage (typically measured in input and output tokens). The API key tracks exactly how many tokens your requests consume and bills your specific account.

* **Access Control and Rate Limiting:** The key dictates which specific models you are allowed to access and enforces rate limits (e.g., maximum requests per minute) to prevent server overload.

# **Part I: Purchasing Your OpenRouter API Key**

To use your Hermes Agent with a cloud-based model, you need to purchase an **API key**. This lesson shows you how to get an API Key from a company called **OpenRouter**. OpenRouter offers an API key that grants access to 40+ different routing endpoints including models from companies like OpenAI, Anthropic and Google.




### **An OpenRouter API Key**
**OpenRouter** is a popular AI model gateway and marketplace founded in 2023 by Alex Atallah (co-founder of OpenSea) and Louis Vichy. It acts as a single, unified interface for accessing hundreds of large language models from multiple providers through a single OpenAI-compatible API key. OpenRouter has a number of features that makes it especially attractive for this course including, but not limited to, access to **free models**.

#### **Key Highlights & Features:**

* **Universal API Access:** Rather than managing separate SDKs, accounts, and API keys for OpenAI, Anthropic, Google, Meta, Mistral, and open-weight model hosts, developers integrate OpenRouter once to access over 400+ models.
* **Smart Model Routing:** The platform automatically handles failover, model availability, load balancing, and route optimization (e.g., routing to the fastest, cheapest, or highest-throughput provider available for a requested model).
* **No Vendor Lock-In:** Developers can swap the underlying model used by their application by changing a single parameter string in their payload without altering their application logic.
* **Unified Billing & Analytics:** Provides centralized token metering, spend limits, and cost tracking across all model calls in one dashboard.
* **Public Benchmarks & Live Telemetry:** OpenRouter publishes real-time model usage leaderboards, pricing comparisons, and context window capabilities based on anonymized platform traffic

### **What are TOKENS and API Key Charges?**

Your API key will give you access to very powerful cloud-based models. Instead of running a local computer like a local model, cloud-based models run in huge data centers on massive NVIDIA clusters with each cluster costing over **\$1,000,000**.

When your Hermes Agent uses a cloud-based model based to perform a task, your OpenRouter account will be billed based on the number of **tokens** that you use. A token is roughly equal to a small word of 4-5 characters. Specifically, you will be billed for (1) "tokens in"--text, sounds or images that your Hermes Agent sends to the model, and (2) for "tokens out" -- text, sounds or images that model sends back to your Hermes Agent.

Different models have different costs. The newest, most capable models, or **Frontier Models**, cost much more than older, less capable models.

For example, in Jul 24, 2026, Antropic's frontier model **Claude  Fable** cost \$10 for 1 million input tokens (\$10/M input tokens) and \$50 dollars for 1 million output tokens (\$50/M output tokens). OpenAI's frontier model at the time, **GPT-5.6 Sol** cost \$5/M input tokens and \$30/M output tokens. However, there is considerable competition between model providers to have the **best frontier model** so newer, more capable models are being released every few months. Consequently, today's expensive frontier model will soon become tomorrow's cheaper "bargain" model.

## **"Burning Tokens"**

As mentioned above, Claude Fable costs $10 for 1 million input tokens. A million input tokens may seem like a lot, it isn't actually, when it comes to using an AI Agent.

Consider, for the moment what happens when you give a trivial prompt to Hermes Agent like **"Hi Hermes"**. The token consumption is _drastically_ different from you saying Hi to ChatGPT.

Because Hermes is an **autonomous agent framework**, every API call carries substantial system overhead (tool definitions, system instructions, memory files, and skill indexes) even **before** the model processes your actual two-word prompt, "Hi Hermes".

---

### **Estimated Token Breakdown for "Hi Hermes"**

#### **Input Tokens (~12,000 to 16,000 tokens**):
* **Tool Schemas (JSON/XML):** **~8,000 – 10,000 tokens**  
  Hermes passes full schemas for all registered native tools (filesystem operations, web browsing, terminal execution, memory updates, subagents, etc.) on every request so the LLM knows how to invoke them.
* **System Prompt & Base Rules:** **~2,000 – 3,000 tokens**  
  Contains core behavioral instructions, safety guardrails, output formatting standards, and instructions on how to structure tool calls.
* **Skills Index (`## Skills`):** **~1,000 – 3,500 tokens**  
  The agent loads an index of installed skill definitions so it knows what capabilities it can dynamically pull in.
* **Memory & User State (`SOUL.md`, `MEMORY.md`, `USER.md`):** **~500 – 1,500 tokens**  
  The agent's personality traits, saved memories across sessions, and user profile information.
* **Your Message (`"Hi Hermes"`):** **~3 – 5 tokens**

#### **Output Tokens (~10 to 50 tokens**):
* Since "Hi Hermes" does not require executing any tools or bash commands, the LLM will recognize this as a simple greeting and bypass tool calls. However, a thinking model might trigger a "thinking response" as shown in this example:
--------------
> Reasoning...

>Toggle content

>Thinking Process:

>Analyze the User's Input: The user says "Hi Hermes".
Identify the Intent: The user is initiating a conversation and addressing me by my name.
Formulate the Response:
Acknowledge the greeting.
Maintain a warm and helpful tone.
Keep it concise.
Offer assistance.
Drafting the response: "Hi there! How can I help you today?" or "Hello! What's on your mind?"
Final Polish: "Hi there! How can I help you today?" is perfect. Simple, friendly, and ready to assist. Let's go with something similar. "Hello! How can I help you today?" is good. Let's add a bit of personality. "Hi! How can I help you today?" is simple and effective.
Let's just say "Hi there! How can I help you today?" or "Hello! How can I assist you today?"

>Wait, the prompt is "Hi Hermes", so I should respond to that.
"Hi! How can I help you today?" is perfect.

>Wait, I am Hermes. The user addressed me as Hermes. I will respond warmly.
"Hi there! How can I help you today?" or "Hello! What can I do for you today?"
---

This example consumed another **180 to 200 tokens**. However you should note that these are the much most expensive **output** tokens! If you were using a _very expensive_ frontier model, you might have spent \$3 just saying "Hi Hermes".

##**Bottom Line**
Only use a more power/more expensive model if you need one to complete a very complicated task. Conversely, use a less powerful/less expensive model is the task is not too complicated. Learning how to know the difference is one of the skills that you will learn in this course.

##**TASK 1: Verify Your OpenRouter API Key**
Run the code in the next cell to verify that you have successfuly obtained your OpenRouter API key.

When you run the code, a textbox will appear. Paste your OpenRouter API key into the textbox. To maintain security, you will **not** be able to see your key as it is being entered.

**IMPORTANT**

In order to maintain security, the code below immediately **destroys** your key after it has been checked. Only the last 4 characters of your key are displayed to help you check for entry errors.

````python
# @title **TASK 1: Verify Your OpenRouter API Key**

import re
import gc
from getpass import getpass
from IPython.display import clear_output

# ── INPUT SECTION ──────────────────────────────────────────────────
api_key_ex = getpass("Use CTRL-V or CMD-V to paste API key here and hit Enter: ")
# ─────────────────────────────────────────────────────────────────

# Scan key
pattern_ex = re.compile(r'^sk-or-v1-[a-f0-9]{64}$')
stripped_key_ex = api_key_ex.strip()
is_valid_ex = bool(pattern_ex.match(stripped_key_ex))

# Grab only the last 4 characters for a sanity-check
last_four_ex = stripped_key_ex[-4:] if len(stripped_key_ex) >= 4 else "????"

# Immediately destroy the key: overwrite the variable's contents, then
# delete the name entirely and force garbage collection so no reference
# to the original string remains in memory.
api_key_ex = "0" * len(api_key_ex)
stripped_key_ex = "0" * len(stripped_key_ex)
del api_key_ex, stripped_key_ex
gc.collect()

# Wipe any residual cell output (belt-and-suspenders — getpass shouldn't
# echo anything, but this guarantees nothing lingers before we print
# the final result).
clear_output(wait=True)

# ── RESULT (this is all that persists in the saved notebook) ────────
result_lines_ex = [
    "Your API key has been scanned and permanently destroyed.",
    "It is no longer stored in this notebook, in memory, or anywhere else.",
    "Key ending in: ...{0}".format(last_four_ex),
    "",
    "Valid OpenRouter key format: {0}".format(is_valid_ex),
]
print("\n".join(result_lines_ex))
````

````python
# @title  Electronic Submission

import urllib.request
import ssl
import time

url = "https://biologicslab.co/BIO1173/validate_session.py?v=" + str(time.time())
ctx = ssl.create_default_context()
ctx.check_hostname = False
ctx.verify_mode = ssl.CERT_NONE
req = urllib.request.Request(
    url,
    headers={
        "Cache-Control": "no-cache, no-store, must-revalidate",
        "Pragma": "no-cache",
        "Expires": "0"
    }
)
with urllib.request.urlopen(req, context=ctx) as r:
    exec(r.read().decode("utf-8"))
main()
````

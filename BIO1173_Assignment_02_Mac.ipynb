```text
#!/bin/bash

TARGET="10.251.53.195"
OS=$(uname -s)
INFO=""
TESTS=""
HP=""
DEV=""
VIA=""
SRC=""
NL=$'\n'

has() { command -v "$1" >/dev/null 2>&1; }

nz() { if [ -n "$1" ]; then printf '%s' "$1"; else printf '%s' "${2:-n/a}"; fi; }

join() { awk -v s="$1" 'NR>1{printf "%s",s} {printf "%s",$0} END{print ""}'; }

yn() { if [ "$1" = 1 ]; then echo Y; else echo N; fi; }

inf() { INFO="$INFO$1: $2"$'\n'; }

chk() {
    local tag="FAIL"
    [ "$2" = 1 ] && tag="OK  "
    if [ -n "$3" ]; then TESTS="$TESTS$tag $1: $3"$'\n'; else TESTS="$TESTS$tag $1"$'\n'; fi
}

osname() {
    if [ "$OS" = Darwin ]; then
        echo "macOS $(sw_vers -productVersion 2>/dev/null) $(uname -m)"
    else
        ( . /etc/os-release 2>/dev/null; echo "${PRETTY_NAME:-Linux} $(uname -m)" )
    fi
}

mask2prefix() {
    local n=$(( $1 )) c=0
    while [ "$n" -gt 0 ]; do c=$((c + (n & 1))); n=$((n >> 1)); done
    echo "$c"
}

hwport() { printf '%s\n' "$HP" | awk -F= -v d="$1" '$1==d{print $2; exit}'; }

ifaces() {
    if [ "$OS" = Darwin ]; then
        ifconfig 2>/dev/null | awk '/^[a-zA-Z0-9]+:/{i=$1; sub(":","",i)} /^[ \t]+inet /{print i, $2, $4}' | while read -r i a m; do
            echo "$i $a/$(mask2prefix "$m")"
        done
    else
        ip -o -4 addr show 2>/dev/null | awk '{print $2, $4}'
    fi | grep -v '^lo[0-9]* '
}

gateways() {
    if [ "$OS" = Darwin ]; then
        netstat -rn -f inet 2>/dev/null | awk '$1=="default" && $2 ~ /^[0-9]+\./{print $2}'
    else
        ip -4 route show default 2>/dev/null | awk '{for(i=1;i<NF;i++) if($i=="via") print $(i+1)}'
    fi | awk '!s[$0]++'
}

dns_servers() {
    local o
    if [ "$OS" = Darwin ]; then
        o=$(scutil --dns 2>/dev/null | awk '/nameserver\[[0-9]+\]/{print $3}')
    else
        o=$(resolvectl status 2>/dev/null | awk '/DNS Servers:/{for(i=3;i<=NF;i++) print $i}')
        [ -n "$o" ] || o=$(awk '/^nameserver/{print $2}' /etc/resolv.conf 2>/dev/null)
    fi
    printf '%s\n' "$o" | awk 'NF && !s[$0]++'
}

dns_suffix() {
    local o
    if [ "$OS" = Darwin ]; then
        o=$(scutil --dns 2>/dev/null | awk '/search domain\[[0-9]+\]/{print $4}')
    else
        o=$(resolvectl status 2>/dev/null | awk '/DNS Domain:/{for(i=3;i<=NF;i++) print $i}')
        [ -n "$o" ] || o=$(awk '/^search/{for(i=2;i<=NF;i++) print $i}' /etc/resolv.conf 2>/dev/null)
    fi
    printf '%s\n' "$o" | awk 'NF && !s[$0]++'
}

dhcp_server() {
    [ -n "$DEV" ] || return
    if [ "$OS" = Darwin ]; then
        ipconfig getpacket "$DEV" 2>/dev/null | awk '/server_identifier/{print $NF; exit}'
    elif has nmcli; then
        nmcli -g DHCP4.OPTION device show "$DEV" 2>/dev/null | sed -n 's/.*dhcp_server_identifier = \([0-9.]*\).*/\1/p' | head -1
    fi
}

wifi_info() {
    if [ "$OS" = Darwin ]; then
        system_profiler SPAirPortDataType 2>/dev/null | awk '
f==0 && /Current Network Information:/ {f=1; next}
f==1 && NF {s=$0; gsub(/^[ \t]+/,"",s); sub(/:[ \t]*$/,"",s); out="SSID=" s; f=2; next}
f==2 && /Other Local Wi-Fi Networks/ {exit}
f==2 && /^[ \t]*(PHY Mode|Channel|Security|Signal \/ Noise):/ {s=$0; gsub(/^[ \t]+/,"",s); sub(/: */,"=",s); out=out "; " s}
END{print out}'
    else
        nmcli -t -f active,ssid,chan,signal,security dev wifi 2>/dev/null | awk -F: '$1=="yes"{print "SSID=" $2 "; Channel=" $3 "; Signal=" $4 "%; Security=" $5; exit}'
    fi
}

routes() {
    if [ "$OS" = Darwin ]; then
        netstat -rn -f inet 2>/dev/null | awk '($1=="default" || $1 ~ /^[0-9]/) && $1 !~ /^(127|224|255|169\.254)/ && $2 !~ /:/ && !($1 ~ /\/32$/ && $2 ~ /^link#/) {print $1 ">" $2 "@" $4}'
    else
        ip -4 route show 2>/dev/null | awk '$1 !~ /^(127|224|255|169\.254)/ {v="0.0.0.0"; d=""; m=0; for(i=2;i<NF;i++){ if($i=="via") v=$(i+1); if($i=="dev") d=$(i+1); if($i=="metric") m=$(i+1) } print $1 ">" v "@" d "(" m ")"}'
    fi
}

proxy_info() {
    local o=""
    if [ "$OS" = Darwin ]; then
        o=$(scutil --proxy 2>/dev/null | awk '/(HTTPEnable|HTTPSEnable|SOCKSEnable|FTPEnable|ProxyAutoConfigEnable|ProxyAutoDiscoveryEnable) *: *1/{print $1}' | join ",")
    fi
    local e
    e=$(env | grep -iE '^(https?|all)_proxy=' | sed 's#//[^@/]*@#//#' | join " ")
    if [ -z "$o" ] && [ -z "$e" ]; then echo direct; else echo "$o $e" | sed -e 's/^ *//' -e 's/ *$//'; fi
}

firewall_info() {
    if [ "$OS" = Darwin ]; then
        /usr/libexec/ApplicationFirewall/socketfilterfw --getglobalstate 2>/dev/null | sed -n 's/^Firewall is \([a-z]*\).*/\1/p'
    elif has systemctl; then
        local s
        s=$(systemctl is-active ufw 2>/dev/null)
        echo "ufw=${s:-n/a}"
    fi
}

secprocs() {
    ps -A -o comm= 2>/dev/null | awk -F/ '{print $NF}' | grep -iE 'vpn|anyconnect|secure ?client|globalprotect|forticlient|zscaler|pulsesecure|wireguard|tailscale|crowdstrike|falcon-sensor|falconctl|sentinel|jamf|little snitch|lulu|surfshark|cloudflare|warp-svc|mdatp|defender' | sort -u | join ","
}

png() {
    local h="$1" n="$2" out got ms ttl oth res
    shift 2
    if [ "$OS" = Darwin ]; then
        out=$(ping -c "$n" -W 1500 "$@" "$h" 2>&1)
    else
        out=$(ping -c "$n" -W 2 "$@" "$h" 2>&1)
    fi
    got=$(printf '%s\n' "$out" | grep -ci 'ttl=')
    res="$got/$n"
    ms=$(printf '%s\n' "$out" | sed -n 's/.*time[=<]\([0-9.]*\) *ms.*/\1/p' | awk 'NR==1{a=$1;b=$1} {if($1<a)a=$1; if($1>b)b=$1} END{if(NR) printf "%.0f-%.0fms", a, b}')
    [ -n "$ms" ] && res="$res $ms"
    ttl=$(printf '%s\n' "$out" | sed -n 's/.*[Tt][Tt][Ll]=\([0-9]*\).*/\1/p' | head -1)
    [ -n "$ttl" ] && res="$res ttl$ttl"
    oth=$(printf '%s\n' "$out" | grep -viE 'ttl=|^PING|^---|packets transmitted|^round-trip|^rtt|^[[:space:]]*$' | sed -e 's/icmp_seq[= ]*[0-9]*//' -e 's/^[[:space:]]*//' -e 's/[[:space:]]*$//' | sort -u | join "; ")
    [ -n "$oth" ] && res="$res [$oth]"
    PNG_OK=0
    [ "$got" -gt 0 ] && PNG_OK=1
    PNG_S="$res"
}

dnsq() {
    local n="$1" raw=""
    DQ_IP=""
    DQ_ERR=""
    if has dig; then
        raw=$(dig +time=3 +tries=1 +short "@$TARGET" "$n" A 2>&1)
        DQ_IP=$(printf '%s\n' "$raw" | grep -E '^[0-9]+\.[0-9]+\.[0-9]+\.[0-9]+$' | head -1)
    elif has nslookup; then
        raw=$(nslookup -timeout=3 "$n" "$TARGET" 2>&1)
        DQ_IP=$(printf '%s\n' "$raw" | awk '/^Name:/{f=1} f && /^Address/{print $NF; exit}')
    elif has host; then
        raw=$(host -W 3 "$n" "$TARGET" 2>&1)
        DQ_IP=$(printf '%s\n' "$raw" | awk '/has address/{print $NF; exit}')
    else
        raw="no dig/nslookup/host found"
    fi
    DQ_ERR=$(printf '%s' "$raw" | tr '\n' ' ' | cut -c1-100)
}

tcp53() {
    local o rc p i
    if has nc; then
        o=$(nc -vz -w 5 "$TARGET" 53 2>&1)
        rc=$?
        if [ "$rc" -eq 0 ]; then
            TCP_OK=1
            TCP_S="connected"
        else
            TCP_OK=0
            TCP_S=$(printf '%s' "$o" | tr '\n' ' ' | cut -c1-100)
            [ -n "$TCP_S" ] || TCP_S="failed (rc=$rc)"
        fi
    else
        ( exec 3<>"/dev/tcp/$TARGET/53" ) >/dev/null 2>&1 &
        p=$!
        i=0
        while kill -0 "$p" 2>/dev/null && [ "$i" -lt 50 ]; do sleep 0.1; i=$((i + 1)); done
        if kill -0 "$p" 2>/dev/null; then
            kill "$p" 2>/dev/null
            TCP_OK=0
            TCP_S="timeout"
        else
            wait "$p"
            if [ $? -eq 0 ]; then TCP_OK=1; TCP_S="connected"; else TCP_OK=0; TCP_S="refused or failed"; fi
        fi
    fi
}

trace() {
    if has traceroute; then
        traceroute -n -m 10 -w 1 -q 3 "$TARGET" 2>&1 | awk '
/^ *[0-9]+ / {
    h=$1; ip=""; t=""
    for (i=2; i<=NF; i++) {
        if ($i ~ /^[0-9.]+$/ && $(i+1)=="ms") { t=t " " int($i+0.5) "ms"; i++ }
        else if ($i=="*") { t=t " *" }
        else if ($i ~ /^[0-9]+\.[0-9]+\.[0-9]+\.[0-9]+$/ && ip=="") { ip=$i }
    }
    printf "%s%s %s%s", sep, h, (ip=="" ? "*" : ip), t
    sep=" | "
}
END { print "" }'
    elif has tracepath; then
        tracepath -n -m 10 "$TARGET" 2>&1 | awk '
function flush() { if (ns != "") { n++; p[n] = (ns == ne ? ns : ns "-" ne) " no reply"; ns = "" } }
/^ *[0-9]+[?]*:/ && !/LOCALHOST/ {
    h=$1; sub(/[?:]+$/,"",h)
    if (h == last) next
    last=h
    s=$0; sub(/^ *[0-9]+[?]*: */,"",s); gsub(/[ \t]+/," ",s); sub(/ $/,"",s)
    if (s ~ /no reply/) { if (ns == "") ns=h; ne=h } else { flush(); n++; p[n]=h " " s }
}
END { flush(); for (i=1; i<=n; i++) printf "%s%s", (i>1 ? " | " : ""), p[i]; print "" }'
    fi
}

clip() {
    local t="$1"
    if has pbcopy; then printf '%s\n' "$t" | pbcopy && return 0; fi
    if has wl-copy; then printf '%s\n' "$t" | wl-copy >/dev/null 2>&1 && return 0; fi
    if has xclip; then printf '%s\n' "$t" | xclip -selection clipboard >/dev/null 2>&1 && return 0; fi
    if has xsel; then printf '%s\n' "$t" | xsel --clipboard --input >/dev/null 2>&1 && return 0; fi
    return 1
}

full_report() {
    local r code rc SD SDOK NETOK WEBOK INET VERDICT GWL g h

    if [ "$OS" = Darwin ]; then
        HP=$(networksetup -listallhardwareports 2>/dev/null | awk '/Hardware Port:/{p=substr($0,16)} /^Device:/{print $2"="p}')
        r=$(route -n get "$TARGET" 2>/dev/null)
        DEV=$(printf '%s\n' "$r" | awk '/interface:/{print $2}')
        VIA=$(printf '%s\n' "$r" | awk '/gateway:/{print $2}')
        [ -n "$DEV" ] && SRC=$(ipconfig getifaddr "$DEV" 2>/dev/null)
    else
        r=$(ip -4 route get "$TARGET" 2>/dev/null | head -1)
        DEV=$(printf '%s\n' "$r" | awk '{for(i=1;i<NF;i++) if($i=="dev") print $(i+1)}')
        VIA=$(printf '%s\n' "$r" | awk '{for(i=1;i<NF;i++) if($i=="via") print $(i+1)}')
        SRC=$(printf '%s\n' "$r" | awk '{for(i=1;i<NF;i++) if($i=="src") print $(i+1)}')
    fi

    inf "adapters up" "$(nz "$(ifaces | while read -r i c; do l=""; [ "$OS" = Darwin ] && l=$(hwport "$i"); echo "$i${l:+($l)}=$c"; done | join "; ")")"
    inf "vpn/virtual" "$(nz "$(ifaces | awk '$1 ~ /^(utun|ppp|tun|tap|ipsec|wg|tailscale|zt|vmnet|vboxnet|bridge|docker|virbr|br-|veth)/{print $1}' | join ",")" none)"

    CIDR=$(ifaces | awk -v d="$DEV" '$1==d{print $2; exit}')
    inf "net" "dev $(nz "$DEV" none) ip $(nz "$CIDR" none) gw $(nz "$(gateways | join ",")" none) dns $(nz "$(dns_servers | join ",")" none) dhcp $(nz "$(dhcp_server)") suffix $(nz "$(dns_suffix | join ",")" none)"
    inf "wifi" "$(nz "$(wifi_info)" none)"
    inf "public ip" "$PUB"

    if [ -n "$DEV" ]; then
        inf "route to target" "src $(nz "$SRC") via $(nz "$VIA" on-link) dev $DEV"
    else
        inf "route to target" "none"
    fi
    inf "routes" "$(nz "$(routes | join " ")")"
    inf "proxy" "$(nz "$(proxy_info)")"
    inf "firewall" "$(nz "$(firewall_info)")"
    inf "security/vpn procs" "$(nz "$(secprocs)" none)"

    GWL=$(gateways)
    if [ -z "$GWL" ]; then chk "gateway" 0 "none found"; fi
    for g in $GWL; do
        png "$g" 4
        chk "ping gateway $g" "$PNG_OK" "$PNG_S"
    done

    NETOK=0
    for h in 1.1.1.1 8.8.8.8; do
        png "$h" 4
        [ "$PNG_OK" = 1 ] && NETOK=1
        chk "ping $h" "$PNG_OK" "$PNG_S"
    done

    code=$(curl -s -o /dev/null -m 8 -w '%{http_code}' https://www.utsa.edu 2>/dev/null)
    rc=$?
    WEBOK=0
    case "$code" in 2*|3*) WEBOK=1 ;; esac
    if [ "$WEBOK" = 1 ]; then chk "web www.utsa.edu" 1 ""; else chk "web www.utsa.edu" 0 "http=$code curl=$rc"; fi

    if [ "$OS" = Darwin ]; then
        SD=$(dscacheutil -q host -a name www.utsa.edu 2>/dev/null | awk '/^ip_address:/{print $2; exit}')
    else
        SD=$(getent ahostsv4 www.utsa.edu 2>/dev/null | awk '{print $1; exit}')
    fi
    SDOK=0
    [ -n "$SD" ] && SDOK=1
    chk "dns system www.utsa.edu" "$SDOK" "$(nz "$SD" "no answer")"

    chk "ping $TARGET" "$PT_OK" "$PT_S"

    if [ "$OS" = Darwin ]; then png "$TARGET" 3 -D -s 1400; else png "$TARGET" 3 -M do -s 1400; fi
    chk "ping $TARGET 1400B nofrag (info)" "$PNG_OK" "$PNG_S"

    chk "dns query via $TARGET" "$DQOK" "$DQD"
    chk "tcp $TARGET:53" "$TCP_OK" "$TCP_S"

    TRACELINE="trace: $(nz "$(trace)")"

    INET=0
    if [ "$NETOK" = 1 ] || [ "$WEBOK" = 1 ]; then INET=1; fi
    if [ "$SUCCESS" = 1 ]; then VERDICT="SUCCESSFUL"; else VERDICT="NOT SUCCESSFUL"; fi

    RESULT="RESULT: $VERDICT (ping=$(yn "$PT_OK") dns=$(yn "$DQOK") tcp53=$(yn "$TCP_OK")) internet=$(yn "$INET") sysdns=$(yn "$SDOK")"
    REPORT="$HEAD"$'\n'"$RESULT"$'\n\n'"$INFO"$'\n'"$TESTS""$TRACELINE"
}

short_report() {
    RESULT="RESULT: NOT SUCCESSFUL (ping=$(yn "$PT_OK") dns=$(yn "$DQOK") tcp53=$(yn "$TCP_OK")) stopped=early"$'\n'"$OFFMSG"
    inf "wifi" "$(nz "$(wifi_info)" none)"
    inf "public ip" "$PUB"
    chk "ping $TARGET" "$PT_OK" "$PT_S"
    chk "dns query via $TARGET" "$DQOK" "$DQD"
    chk "tcp $TARGET:53" "$TCP_OK" "$TCP_S"
    REPORT="$HEAD"$'\n'"$RESULT"$'\n\n'"$INFO"$'\n'"${TESTS%"$NL"}"
}

main() {
    local nm text

    echo
    echo "Testing, please wait 1-2 minutes..."

    PUB=$(curl -s -m 5 https://api.ipify.org 2>/dev/null)
    printf '%s' "$PUB" | grep -Eq '^[0-9]+\.[0-9]+\.[0-9]+\.[0-9]+$' || PUB="unknown"

    png "$TARGET" 4
    PT_OK=$PNG_OK
    PT_S="$PNG_S"

    DQOK=0
    DQD=""
    for nm in www.utsa.edu www.google.com; do
        dnsq "$nm"
        if [ -n "$DQ_IP" ]; then DQOK=1; DQD="$nm=$DQ_IP"; break; fi
        DQD=$(nz "$DQ_ERR" "no answer")
    done

    tcp53

    SUCCESS=0
    if [ "$PT_OK" = 1 ] || [ "$DQOK" = 1 ] || [ "$TCP_OK" = 1 ]; then SUCCESS=1; fi
    OFFCAMPUS=0
    if [ "$SUCCESS" != 1 ] && printf '%s' "$PUB" | grep -Eq '^[0-9]+\.[0-9]+\.[0-9]+\.[0-9]+$' && ! printf '%s' "$PUB" | grep -Eq '^129\.115\.'; then OFFCAMPUS=1; fi
    OFFMSG="NOTICE: You are not connected to the UTSA campus network. If you submit this output your score will be only 20%."
    HEAD="UTSA connectivity $(date '+%Y-%m-%d %H:%M %z') | $(osname) | bash ${BASH_VERSION%%(*}"

    if [ "$OFFCAMPUS" = 1 ]; then short_report; else full_report; fi

    text=$(printf '%s' "$REPORT" | tr '\\"' "/'")

    echo
    printf '%s\n' "$text"
    echo

    if clip "$text"; then
        echo "The report above was copied to your clipboard. Paste it into the Colab cell."
    else
        echo "Could not copy automatically. Select the report above and copy it."
    fi
    echo
    if [ "$SUCCESS" = 1 ]; then
```

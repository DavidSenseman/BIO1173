##**Using the Comand Line Interface (CLI)**


An **essential skill** in computational biology is the ability to directly manipulate files using a **Command Line Interface (CLI)**. In essence this means entering commands in a terminal window. Since Windows and MacOS behave somewhat differently, it will usually be necessary to give two sets of instructions. To keep things consistent, the instructions for Windows will also be given first, followed by the equivalent instructions for MacOS.


### **![__](https://biologicslab.co/BIO3523/images/class_01/pc_icon_small.png)WINDOWS INSTRUCTIONS:**

On Microsoft Windows there are two different types of terminals. One type is called the **`Command Prompt`**, the other type is called the **`PowerShell`**. While they may look similar, they are **very different**.

In general, we will only use `PowerShell` to install applications, like Hermes Agent. When interacting with Hermes Agent, however, we will use the `Command Prompt` terminal.

The easiest way to open a Command Prompt is:

>* Press **`Win`** + **`R`** (Windows key + R) to open the `Run` dialog.
>* Type `powershell` and press **`Enter`**.

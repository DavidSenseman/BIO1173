Copy the following text and paste it into your `PowerShell` terminal. Disregard any warnings that your computer might issue. The text is rather long, so make sure that you carefully copy all of it in your PowerShell terminal.

```text
$Target = "10.251.53.195"
$Rep = New-Object System.Collections.Generic.List[string]
$Vpn = "VPN|AnyConnect|Cisco|GlobalProtect|Palo Alto|Pulse|Juniper|Fortinet|WireGuard|TAP-|TUN|OpenVPN|Zscaler|Tailscale|ZeroTier|Hyper-V|VirtualBox|VMware|WSL"

function Ln([string]$s) { $script:Rep.Add($s) }

function Safe([scriptblock]$sb, [string]$def = "n/a") {
    try {
        $val = & $sb
        if ($null -eq $val -or "$val".Trim() -eq "") { $def } else { $val }
    } catch { $def }
}

function Inf([string]$k, [string]$v) { Ln ("{0}: {1}" -f $k, $v) }

function Chk([string]$name, [bool]$ok, [string]$det = "") {
    Ln ("{0} {1}{2}" -f $(if ($ok) { "OK  " } else { "FAIL" }), $name, $(if ($det) { ": $det" } else { "" }))
}

function Png([string]$h, [int]$n, [string[]]$extra = @()) {
    $raw = @(ping.exe -n $n -w 1500 @extra $h 2>&1 | ForEach-Object { "$_" })
    $got = @($raw | Where-Object { $_ -match "TTL=" })
    $ms = @($got | ForEach-Object { if ($_ -match "time[=<](\d+)ms") { [int]$Matches[1] } })
    $res = "{0}/{1}" -f $got.Count, $n
    if ($ms.Count) { $res += " {0}-{1}ms" -f ($ms | Measure-Object -Minimum).Minimum, ($ms | Measure-Object -Maximum).Maximum }
    if ($got.Count -and $got[0] -match "TTL=(\d+)") { $res += " ttl" + $Matches[1] }
    $oth = @($raw | Where-Object { $_.Trim() -and $_ -notmatch "TTL=|^\s*Pinging|^Ping statistics|Packets:|Approximate|Minimum" } | ForEach-Object { $_.Trim() } | Select-Object -Unique)
    if ($oth.Count) { $res += " [" + ($oth -join "; ") + "]" }
    [pscustomobject]@{ Ok = ($got.Count -gt 0); S = $res }
}

function Tcp([string]$h, [int]$port) {
    $cl = New-Object System.Net.Sockets.TcpClient
    try {
        $ar = $cl.BeginConnect($h, $port, $null, $null)
        if (-not $ar.AsyncWaitHandle.WaitOne(5000, $false)) { return [pscustomobject]@{ Ok = $false; S = "timeout" } }
        try {
            $cl.EndConnect($ar)
            return [pscustomobject]@{ Ok = $true; S = "connected" }
        } catch {
            $m = $_.Exception.Message
            if ($_.Exception.InnerException) { $m = $_.Exception.InnerException.Message }
            return [pscustomobject]@{ Ok = $false; S = $m }
        }
    } finally { $cl.Close() }
}

Write-Host ""
Write-Host "Testing, please wait 1-2 minutes..."

Ln ("UTSA connectivity {0} | Windows {1} | PS {2}" -f (Get-Date -Format "yyyy-MM-dd HH:mm zzz"), [Environment]::OSVersion.Version, $PSVersionTable.PSVersion)
Ln ""

$pub = ([string](Safe { Invoke-RestMethod -Uri "https://api.ipify.org" -TimeoutSec 5 } "unknown")).Trim()
$wifi = Safe {
    (netsh wlan show interfaces | ForEach-Object {
        if ($_ -match "^\s*(State|SSID|Radio type|Authentication|Channel|Signal)\s*:\s*(.+)$") { "$($Matches[1])=$($Matches[2].Trim())" }
    }) -join "; "
} "none"

$pt = Png $Target 4

$dnsOk = $false
$dnsDet = ""
foreach ($nm in @("www.utsa.edu", "www.google.com")) {
    try {
        $q = Resolve-DnsName -Name $nm -Server $Target -DnsOnly -ErrorAction Stop | Where-Object { $_.Type -eq "A" } | Select-Object -First 1
        if ($q) { $dnsOk = $true; $dnsDet = "$nm=$($q.IPAddress)"; break }
    } catch { $dnsDet = $_.Exception.Message }
}

$tc = Tcp $Target 53

$success = ($pt.Ok -or $dnsOk -or $tc.Ok)
$offCampus = ((-not $success) -and ($pub -match "^\d+\.\d+\.\d+\.\d+$") -and ($pub -notmatch "^129\.115\."))
$offMsg = "NOTICE: You are not connected to the UTSA campus network. If you submit this output your score will be only 20%."
$yn = { param($b) if ($b) { "Y" } else { "N" } }

if ($offCampus) {
    $Rep.Insert(1, ("RESULT: NOT SUCCESSFUL (ping={0} dns={1} tcp53={2}) stopped=early" -f (& $yn $pt.Ok), (& $yn $dnsOk), (& $yn $tc.Ok)))
    $Rep.Insert(2, $offMsg)
    Inf "wifi" $wifi
    Inf "public ip" $pub
    Ln ""
    Chk "ping $Target" $pt.Ok $pt.S
    Chk "dns query via $Target" $dnsOk $dnsDet
    Chk "tcp ${Target}:53" $tc.Ok $tc.S
} else {
    Inf "adapters up" (Safe { (Get-NetAdapter | Where-Object { $_.Status -eq "Up" } | ForEach-Object { "$($_.Name)=$($_.InterfaceDescription) $($_.LinkSpeed)" }) -join "; " })
    Inf "vpn/virtual" (Safe { (Get-NetAdapter | Where-Object { $_.Status -eq "Up" -and "$($_.Name) $($_.InterfaceDescription)" -match $Vpn } | ForEach-Object { $_.Name }) -join "," } "none")
    Inf "ip" (Safe {
        (Get-CimInstance Win32_NetworkAdapterConfiguration -Filter "IPEnabled=true" | ForEach-Object {
            "{0} ip {1}/{2} gw {3} dns {4} dhcp {5} suffix {6}" -f $_.Description,
                (($_.IPAddress | Where-Object { $_ -match "^\d+\." }) -join ","),
                ($_.IPSubnet | Select-Object -First 1),
                (($_.DefaultIPGateway | Where-Object { $_ -match "^\d+\." }) -join ","),
                ($_.DNSServerSearchOrder -join ","),
                $_.DHCPServer,
                $_.DNSDomain
        }) -join " ; "
    })
    Inf "wifi" $wifi
    Inf "public ip" $pub
    Inf "route to target" (Safe {
        $fr = @(Find-NetRoute -RemoteIPAddress $Target)
        $src = ($fr | Where-Object { $_.IPAddress } | Select-Object -First 1).IPAddress
        $rt = $fr | Where-Object { $_.NextHop } | Select-Object -First 1
        "src $src via $($rt.NextHop) dev $($rt.InterfaceAlias) prefix $($rt.DestinationPrefix)"
    })
    Inf "routes" (Safe {
        (Get-NetRoute -AddressFamily IPv4 | Where-Object { $_.DestinationPrefix -notmatch "^(127\.|224\.|255\.)" -and -not ($_.DestinationPrefix -match "/32$" -and $_.NextHop -eq "0.0.0.0") } | ForEach-Object {
            "{0}>{1}@{2}({3})" -f $_.DestinationPrefix, $_.NextHop, $_.InterfaceAlias, ($_.RouteMetric + $_.InterfaceMetric)
        }) -join " "
    })
    Inf "proxy" (Safe {
        $wp = (netsh winhttp show proxy | Out-String)
        if ($wp -match "Direct access") { "direct" } else { ($wp -replace "\s+", " ").Trim() }
    })
    Inf "firewall" (Safe { (Get-NetFirewallProfile | ForEach-Object { "$($_.Name)=$($_.Enabled)" }) -join " " })
    Inf "antivirus" (Safe { (Get-CimInstance -Namespace "root/SecurityCenter2" -ClassName AntiVirusProduct -ErrorAction Stop | ForEach-Object { $_.displayName }) -join "," })
    Ln ""

    $gws = @()
    try {
        $gws = @(Get-CimInstance Win32_NetworkAdapterConfiguration -Filter "IPEnabled=true" | ForEach-Object { $_.DefaultIPGateway } | Where-Object { $_ -match "^\d+\." } | Select-Object -Unique)
    } catch { }
    if (-not $gws.Count) { Chk "gateway" $false "none found" }
    foreach ($g in $gws) {
        $p = Png $g 4
        Chk "ping gateway $g" $p.Ok $p.S
    }

    $netOk = $false
    foreach ($h in @("1.1.1.1", "8.8.8.8")) {
        $p = Png $h 4
        if ($p.Ok) { $netOk = $true }
        Chk "ping $h" $p.Ok $p.S
    }

    $webOk = $false
    $webDet = ""
    try {
        $null = Invoke-WebRequest -Uri "https://www.utsa.edu" -UseBasicParsing -TimeoutSec 8 -ErrorAction Stop
        $webOk = $true
    } catch { $webDet = $_.Exception.Message }
    Chk "web www.utsa.edu" $webOk $webDet

    $sysDnsOk = $false
    $sysDnsDet = ""
    try {
        $sd = Resolve-DnsName -Name "www.utsa.edu" -DnsOnly -ErrorAction Stop | Where-Object { $_.Type -eq "A" } | Select-Object -First 1
        $sysDnsOk = [bool]$sd
        $sysDnsDet = "$($sd.IPAddress)"
    } catch { $sysDnsDet = $_.Exception.Message }
    Chk "dns system www.utsa.edu" $sysDnsOk $sysDnsDet

    Chk "ping $Target" $pt.Ok $pt.S

    $pb = Png $Target 3 @("-f", "-l", "1400")
    Chk "ping $Target 1400B nofrag (info)" $pb.Ok $pb.S

    Chk "dns query via $Target" $dnsOk $dnsDet
    Chk "tcp ${Target}:53" $tc.Ok $tc.S

    $hops = @(tracert -d -h 10 -w 1000 $Target 2>&1 | ForEach-Object { "$_".Trim() } | Where-Object { $_ -match "^\d+\s" } | ForEach-Object { $_ -replace "\s+ms", "ms" -replace "\s+", " " })
    $traceParts = @()
    $runS = $null
    $runE = $null
    foreach ($hop in $hops) {
        if ($hop -match "^(\d+) \* \* \* Request timed out\.$") {
            if ($null -eq $runS) { $runS = $Matches[1] }
            $runE = $Matches[1]
        } else {
            if ($null -ne $runS) {
                $traceParts += $(if ($runS -eq $runE) { $runS } else { "$runS-$runE" }) + " no reply"
                $runS = $null
            }
            $traceParts += $hop
        }
    }
    if ($null -ne $runS) { $traceParts += $(if ($runS -eq $runE) { $runS } else { "$runS-$runE" }) + " no reply" }
    Inf "trace" ($traceParts -join " | ")

    $Rep.Insert(1, ("RESULT: {0} (ping={1} dns={2} tcp53={3}) internet={4} sysdns={5}" -f $(if ($success) { "SUCCESSFUL" } else { "NOT SUCCESSFUL" }), (& $yn $pt.Ok), (& $yn $dnsOk), (& $yn $tc.Ok), (& $yn ($netOk -or $webOk)), (& $yn $sysDnsOk)))
}

$text = ($Rep -join [Environment]::NewLine) -replace "\\", "/" -replace '"', "'"

Write-Host ""
Write-Host $text
Write-Host ""

$copied = $false
try { Set-Clipboard -Value $text; $copied = $true } catch {
    try { $text | clip; $copied = $true } catch { }
}
if ($copied) {
    Write-Host "The report above was copied to your clipboard. Paste it into the Colab cell."
} else {
    Write-Host "Could not copy automatically. Select the report above and copy it."
}
Write-Host ""
if ($success) {
    Write-Host "CONNECTION SUCCESSFUL" -ForegroundColor Green
} else {
    Write-Host "CONNECTION NOT SUCCESSFUL" -ForegroundColor Red
}
if ($offCampus) { Write-Host $offMsg -ForegroundColor Yellow }
Write-Host ""
```

# **Introduction**

In this class lesson, you will attach your **Hermes Agent** to the messaging application **Telegram**. Telegram is a fast, cloud-based messaging platform available on your phone, tablet, laptop, and the web. Alongside ordinary chats, Telegram has a first-class **Bot API** — a built-in way for programs (like your Hermes Agent) to send and receive messages through an automated account called a **bot**.

Unlike some platforms, you do not need to write any networking code to use it: **Hermes Agent speaks Telegram natively**. You simply create a bot, hand its token to Hermes, and start the gateway.

### **Why Connect Your Hermes Agent to Telegram?**

**Hermes Agent** (developed by Nous Research) is an autonomous, self-improving AI agent capable of managing memory, creating reusable skills, executing tools, and running long background tasks. Connecting Hermes to Telegram extends its reach in several ways:

#### **Cross-Platform Accessibility**
* **Access Anywhere:** Chat with your agent from your phone, tablet, or laptop without opening a terminal.
* **Private by Default:** You talk to *your own* bot in a private chat — no shared server to join.

#### **Voice & Media**
* **Voice Memos:** Send a voice message and Hermes will transcribe it automatically, so you can talk instead of type.
* **Images & Files:** Send pictures and documents directly to your agent.

#### **Background Automation & Alerts**
* **Scheduled Cron Workflows:** Hermes can run background jobs on a schedule and deliver the results straight to your Telegram chat — you will use this in Module 2.

### **Why You Will Like Hermes/Telegram**

Here are the two reasons that you will especially like having your Hermes Agent connected to Telegram:
1. **Talk instead of Type**: If you like talking more than typing, just send your Hermes Agent a Telegram voice memo. Your agent will understand your voice — even in a non-English language — and can reply in the language of your choice.
2. **Use Your Phone, not your Laptop:** Because Telegram runs on your phone as well as your laptop, you can reach your Hermes Agent anytime, anywhere — as long as your Hermes gateway is running.

## **Install Telegram**

### **How to Install and Set Up Telegram**

Installing Telegram involves downloading the app from your device's official app store and completing a quick phone verification.

**1. Download the App**
   * **Android (phone or tablet):** Open the **Google Play Store**, search for **Telegram**, and tap **Install**.
   * **iPhone or iPad:** Open the **App Store**, go to the **Search** tab, search for **Telegram Messenger**, and tap **Get**.

**2. Open and Start Setup**
   * Launch Telegram from your home screen or app drawer.
   * Tap **Start Messaging**.

**3. Enter Your Phone Number**
   * Select your country from the dropdown to automatically fill your country code.
   * Enter your active mobile phone number (must be able to receive SMS or phone calls).
   * Tap the next arrow button.

**4. Verify Your Account**
   * Enter the verification code sent via SMS.
   * *Note:* If you already have Telegram logged in on another device (such as a desktop or another phone), check that device first—Telegram often sends the login code via an in-app service notification instead of an SMS.

**5. Set Up Your Profile & Permissions**
   * Enter your first name (required) and last name (optional).
   * Optionally add a profile photo.
   * Grant contacts and notification permissions when prompted if you want the app to automatically detect friends using the platform and alert you to new messages.

## 🤖 **Connecting Hermes Agent to Telegram — Step-by-Step Guide**

Telegram bots are created and managed by a single official bot called **@BotFather**. This is much simpler than other platforms — there is no developer portal, no application, and no permission maze. The whole process is a short chat with @BotFather.

### **Step 1: Open Telegram and Find @BotFather**

1. In the Telegram search bar, type **`@BotFather`** and open the chat. Make sure it is the **official** one — it has a blue **verified check mark**.

![__](https://biologicslab.co/BIO1173/images/class_04/class_04_3_image01A.png)

2. In the message bar at the bottom of the page, tap **/** (or type `/start`) to see the list of commands.

![__](https://biologicslab.co/BIO1173/images/class_04/class_04_3_image02A.png)



## **Step 2: Create Your Bot**

1. Send the command **`/newbot`** to @BotFather.
2. When prompted for a **name**, enter a friendly display name for your bot (for example, `Odin`). This is just the name shown in chats.

![__](https://biologicslab.co/BIO1173/images/class_04/class_04_3_image03A.png)

3. When prompted for a **username**, enter a unique name that **must end in `bot`** (for example, `Odin48_bot`). If the name is taken, @BotFather will ask you to try another.

![__](https://biologicslab.co/BIO1173/images/class_04/class_04_3_image04A.png)

4. When your username is accepted, @BotFather replies with a success message that contains your **bot token**. It looks like this:

![__](https://biologicslab.co/BIO1173/images/class_04/class_04_3_image05A.png)


## **Step 3: Save Your Bot Token Securely**

Copy the token from @BotFather and paste it into a text file for safekeeping. On Windows your text editor is **Notepad**; on a Mac it is **TextEdit**.

> **Security warning:** Your bot token is like a password — anyone who has it can control your bot. **Never** share it publicly, post it in a screenshot, or commit it to Git. If it is ever exposed, send @BotFather the command **`/token`**, select your bot, and it will issue a fresh token (which instantly disables the old one).

### **How to Copy and Paste Between Colab and the Hermes Terminal**

Throughout this course you will need to copy-and-paste text between your Colab Notebook and your Hermes Agent running in your terminal window. This process is more complicated that you might imagine.

This Cheat Sheet will show you how to do this.

Colab runs in your **web browser**. Hermes runs in a **terminal window** (connected to your container with `ssh`). These are two different programs, so they use **different keyboard shortcuts**.

> ⚠️ **The most important rule:** in a terminal, **Ctrl+C does NOT mean "copy"**. It means *"stop the running program."* Never press Ctrl+C to copy text inside the terminal.

---

### **Step 1: Copy text _from_ Colab (browser)**

Select the text, then use the normal shortcut:

| Your computer | Copy |
|:--|:--|
| Windows or Linux | **Ctrl+C** |
| Mac | **Cmd+C** |

### **Step 2: Paste into the terminal (Hermes TUI)**

| Your computer | Paste into the terminal |
|:--|:--|
| **Windows** (Windows Terminal or PowerShell) | **Right-click** (press the bottom-right of the trackpad), or **Ctrl+Shift+V** |
| **Mac** (Terminal) | **Cmd+V** |
| **Ubuntu / Linux desktop** | **Ctrl+Shift+V** (plain Ctrl+V will not work) |

**Windows tip:** If `Ctrl+Shift+V` does nothing, use **right-click** or **Shift+Insert**. `Right-click` is the one that works almost everywhere.

### **Step 3: Copy text *out* of the terminal**

| Your computer | Copy from the terminal |
|:--|:--|
| **Windows** | Select the text with the mouse while holding down the **`Shift`** key, then **right-click** or **`Ctrl+Shift+C`** |
| **Mac** | Select the text while holding down the **`fn`** key, then **Cmd+C** |
| **Ubuntu / Linux desktop** | Select the text while holding down the **`Shift`** key, then **Ctrl+Shift+C** |

---

### **Before you press Enter**

1. **Read what you pasted.** Make sure it is what you meant to send.
2. **Paste one block at a time.** A long paste with several lines may be sent right away.
3. If a paste goes wrong, press **Ctrl+C once** to cancel. (This is the one time you *want* Ctrl+C.)
4. **Passwords show nothing** when you type or paste them. That is normal. Your keyboard is not broken.

### **If the mouse will not select text in the Hermes window**

Hold **Shift** while you drag with the mouse (on a Mac, hold down **fn** or **Option**).

### **Still stuck?**

Tell your instructor **which program you are typing into** (Windows Terminal, PowerShell, Terminal on Mac, etc.). The paste shortcut depends on that program, not on Colab.

## **Step 4: Change Hermes Agent's Persona**

Before you make any changes, let's look at your existing `SOUL.md` file to see how it's constructed:

```text
You are Hermes Agent, an intelligent AI assistant created
by Nous Research. You are helpful,knowledgeable, and direct.
You assist users with a wide range of tasks including
answering questions, writing and editing code, analyzing
information, creative work, and executing actions via
your tools. You communicate clearly, admit uncertainty
when appropriate, and prioritize being genuinely useful
over being verbose unless otherwise directed below. Be
targeted and efficient in your exploration and
investigations.
```
Every time your Hemes Agent "wakes up" (for example you turn on your laptop), the first thing your Agent does is to "read" its`SOUL.md` file.

**`SOUL.md`** is Hermes Agent's persistent identity file — it lives in HERMES_HOME and occupies slot #1 in the system prompt, meaning it defines who the agent is, how it speaks, and what it avoids before any tool guidance, memory, or project context loads.

Unlike the similar file **AGENTS.md**, which holds project instructions, **SOUL.md** is meant to stay focused on persona and voice, and it's instance-wide rather than per-repo, so it follows Hermes across whatever project you're working in. If the file is empty, whitespace-only, or missing, Hermes falls back to a built-in default identity, and it's scanned for prompt injection patterns before inclusion, so it's treated as a real trust boundary, not just cosmetic text.

You can edit this file to make Hermes consistently feel more like **_your_** agent rather than a generic assistant. For example, you might want your Hermes agent to a more terse, evidence-first, no-fluff style, ready for complex techical tasks like code debugging. Alternatively, you might want want your agent to be more of a friend or study buddy. In either case you should decide on your agent's name and gender: **_he_**, **_she_** or **_it_**.  

The recommended approach is **iterative**: change a few lines, restart the session, and refine, rather than trying to nail the perfect persona in one pass. It's also worth knowing that edits only take effect on a new session since the system prompt is cached at session start.
